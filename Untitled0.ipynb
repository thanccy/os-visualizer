```python

import pandas as pd
import matplotlib.pyplot as plt
from collections import deque


def round_robin(processes, quantum, context_overhead=0):
    # Process format: [name, arrival_time, burst_time]
    data = {
        p[0]: {
            "arrival": p[1],
            "burst": p[2],
            "remaining": p[2],
            "first_start": None,
            "completion": None
        }
        for p in processes
    }

    processes = sorted(processes, key=lambda p: p[1])
    ready = deque()
    timeline = []
    time = 0
    i = 0
    last_process = None
    switches = 0

    while i < len(processes) or ready:

        # Add processes that have arrived
        while i < len(processes) and processes[i][1] <= time:
            ready.append(processes[i][0])
            i += 1

        # CPU is idle until the next process arrives
        if not ready:
            next_arrival = processes[i][1]
            if time < next_arrival:
                timeline.append(("Idle", time, next_arrival))
                time = next_arrival
            continue

        current = ready.popleft()

        # Add context switching overhead when processes change
        if last_process is not None and current != last_process:
            switches += 1
            if context_overhead > 0:
                start = time
                time += context_overhead
                timeline.append(("CS", start, time))

                # Processes arriving during the context switch
                while i < len(processes) and processes[i][1] <= time:
                    ready.append(processes[i][0])
                    i += 1

        p = data[current]

        if p["first_start"] is None:
            p["first_start"] = time

        start = time
        run_time = min(quantum, p["remaining"])
        time += run_time
        p["remaining"] -= run_time
        timeline.append((current, start, time))

        # Add newly arrived processes before re-queuing current
        while i < len(processes) and processes[i][1] <= time:
            ready.append(processes[i][0])
            i += 1

        if p["remaining"] > 0:
            ready.append(current)
        else:
            p["completion"] = time

        last_process = current

    # Calculate performance metrics
    results = []

    for name, p in data.items():
        completion = p["completion"]
        turnaround = completion - p["arrival"]
        waiting = turnaround - p["burst"]
        response = p["first_start"] - p["arrival"]

        results.append([
            name, p["arrival"], p["burst"], completion,
            turnaround, waiting, response
        ])

    columns = [
        "Process", "Arrival", "Burst", "Completion",
        "Turnaround", "Waiting", "Response"
    ]
    df = pd.DataFrame(results, columns=columns)

    averages = {
        "Average Turnaround": df["Turnaround"].mean(),
        "Average Waiting": df["Waiting"].mean(),
        "Average Response": df["Response"].mean()
    }

    total_burst = sum(p[2] for p in processes)
    total_overhead = switches * context_overhead
    cpu_efficiency = (
        total_burst / (total_burst + total_overhead) * 100
        if total_burst + total_overhead > 0 else 0
    )

    return df, timeline, averages, switches, total_overhead, cpu_efficiency


def show_gantt(timeline, title):
    fig, ax = plt.subplots(figsize=(12, 3))

    colors = {
        "P1": "skyblue",
        "P2": "lightgreen",
        "P3": "orange",
        "P4": "plum",
        "CS": "tomato",
        "Idle": "lightgray"
    }

    for label, start, end in timeline:
        ax.barh(
            0, end - start, left=start, height=0.5,
            color=colors.get(label, "gray"),
            edgecolor="black"
        )
        ax.text(
            (start + end) / 2, 0, label,
            ha="center", va="center", fontsize=9
        )
        ax.text(start, -0.38, str(start), ha="center", fontsize=8)

    if timeline:
        ax.text(timeline[-1][2], -0.38, str(timeline[-1][2]),
                ha="center", fontsize=8)

    ax.set_title(title)
    ax.set_xlabel("Time (ms)")
    ax.set_yticks([])
    ax.set_xlim(0, max(item[2] for item in timeline) + 1)
    ax.grid(axis="x", linestyle="--", alpha=0.4)
    plt.tight_layout()
    plt.show()


def run_activity(title, processes, quantums, overhead):
    print("\n" + "=" * 65)
    print(title)
    print("=" * 65)

    comparisons = []

    for quantum in quantums:
        df, timeline, averages, switches, total_overhead, efficiency = (
            round_robin(processes, quantum, overhead)
        )

        print(f"\nTime Quantum = {quantum} ms")
        print(df.to_string(index=False))

        print(f"\nAverage Turnaround Time: "
              f"{averages['Average Turnaround']:.2f} ms")
        print(f"Average Waiting Time: "
              f"{averages['Average Waiting']:.2f} ms")
        print(f"Average Response Time: "
              f"{averages['Average Response']:.2f} ms")
        print(f"Context Switches: {switches}")
        print(f"Total Switching Overhead: {total_overhead} ms")
        print(f"CPU Efficiency: {efficiency:.2f}%")

        show_gantt(timeline, f"{title} - Quantum {quantum} ms")

        comparisons.append({
            "Quantum (ms)": quantum,
            "Avg Turnaround (ms)": round(
                averages["Average Turnaround"], 2),
            "Avg Waiting (ms)": round(
                averages["Average Waiting"], 2),
            "Avg Response (ms)": round(
                averages["Average Response"], 2),
            "Context Switches": switches,
            "Overhead (ms)": total_overhead,
            "CPU Efficiency (%)": round(efficiency, 2)
        })

    comparison = pd.DataFrame(comparisons)
    print("\nCOMPARISON TABLE")
    print(comparison.to_string(index=False))
    return comparison


# ACTIVITY 1: No context-switching overhead
municipal_processes = [
    ["P1", 0, 5],  # Water-quality monitoring
    ["P2", 1, 3],  # Waste-collection scheduling
    ["P3", 2, 4],  # Traffic-pollution analysis
    ["P4", 3, 2]   # Public energy-use monitoring
]

activity1 = run_activity(
    "ACTIVITY 1: Municipal Sustainability Services",
    municipal_processes,
    quantums=[2, 3],
    overhead=0
)


# ACTIVITY 2: One millisecond context-switching overhead
emergency_processes = [
    ["P1", 0, 5],  # Flood-warning analysis
    ["P2", 1, 4],  # Air-quality alert generation
    ["P3", 2, 3],  # Emergency evacuation-route update
    ["P4", 3, 2]   # Renewable-grid monitoring
]

activity2 = run_activity(
    "ACTIVITY 2: Climate Emergency Response",
    emergency_processes,
    quantums=[2, 4],
    overhead=1
)
```
